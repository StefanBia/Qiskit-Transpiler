```python
from qiskit import QuantumCircuit, ClassicalRegister, QuantumRegister
from qiskit.circuit.library import RYGate, RZGate, CXGate

qc = QuantumCircuit(4)

new_qc = QuantumCircuit(2)

qc.append(RYGate(1.234), [0])
qc.append(CXGate(), [0, 1])
qc.append(RZGate(0.567), [1])

for instrunction in qc.data:
    print(instrunction)
    print(instrunction.operation)
    for qubit in instrunction.qubits:
        idx = qc.find_bit(qubit).index
        print(idx)

for instruction in qc.data:
    op = instruction.operation
    qubits = [qc.find_bit(qubit).index for qubit in instruction.qubits]
    new_qc.append(op, qubits)

qc.draw(output='mpl')
```
