```python
from qiskit import QuantumCircuit, ClassicalRegister, QuantumRegister
from qiskit.circuit.library import RYGate, RZGate, CXGate

qubits = QuantumRegister(4, 'q')
clbits = ClassicalRegister(2, 'c')
qc = QuantumCircuit(qubits, clbits)

new_qc = QuantumCircuit(4, 2)

qc.append(RYGate(1.234), [0])
qc.append(CXGate(), [0, 1])
qc.append(RZGate(0.567), [1])

qc.x(0)
qc.measure(0, 0)
# for instruction in qc.data:
#     print(instruction)
#     print(instruction.operation)
#     for qubit in instruction.qubits:
#         idx = qc.find_bit(qubit).index
#         print(idx)

for instruction in qc.data:
    op = instruction.operation
    qubits = [qc.find_bit(qubit).index for qubit in instruction.qubits]
    cbits = [qc.find_bit(cbit).index for cbit in instruction.clbits]
    print(f"Operation: {op}, Qubits: {qubits}, Classical bits: {cbits}")
    
    new_qc.append(op, qubits, cbits)

qc.draw(output='mpl')
```
